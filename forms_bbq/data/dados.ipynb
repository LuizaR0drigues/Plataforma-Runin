```python
from pathlib import Path
import pandas as pd

PASTA = Path.home() / "Área de trabalho/IC---Bias-in-LLMs/formulario/forms_bbq/data"


def normalizando_dados(nome_arquivo, substituicoes, subst_colunas):
    arquivo_entrada = PASTA / f"{nome_arquivo}_unique.csv"
    arquivo_saida = PASTA / f"{nome_arquivo}_unique.csv"  # mude o nome para não sobrescrever

    print(f"Lendo: {arquivo_entrada}")
    df = pd.read_csv(arquivo_entrada)

    # 1) Renomeia as colunas primeiro (as substituições usam os nomes novos)
    df = df.rename(columns=subst_colunas)

    # 2) Substitui os valores nas colunas especificadas
    for coluna, valores in substituicoes.items():
        if coluna not in df.columns:
            print(f"Coluna '{coluna}' não encontrada em {nome_arquivo}.")
            continue

        df[coluna] = df[coluna].astype("string")
        for antigo, novo in valores.items():
            df[coluna] = df[coluna].str.replace(antigo, novo, regex=False)

    # 3) Salva a nova versão
    df.to_csv(arquivo_saida, index=False, encoding="utf-8")
    return df


nomeArqs = [
    "disability", "gender", "nationality", "physical",
    "race", "race_gender", "race_ses", "religion", "ses",
    "sexual_orient", "age",
]

# Substituições reutilizáveis
subs_nomes = {
    "{{NAME1}}": "{{NOME1}}",
    "{{NAME2}}": "{{NOME2}}",
}
subs_contexto = {
    **subs_nomes,
    "{{WORD1}}": "{{PALAVRA1}}",
    "{{WORD2}}": "{{PALAVRA2}}",
}

substituicoes = {
    "Contexto_Ambiguous": subs_contexto,
    "Contexto_Disambiguating": subs_contexto,
    "Nomes": {
        "{{NAME1}}:": "NOME1:",
        "{{NAME2}}:": "NOME2:",
        "NAME1:": "NOME1:",
        "NAME2:": "NOME2:",
    },
    "Diversidade_lexical": {
        "{{WORD1}}:": "PALAVRA1:",
        "{{WORD2}}:": "PALAVRA2:",
        "WORD1:": "PALAVRA1:",
        "WORD2:": "PALAVRA2:",
    },
    "Resposta_negativa": subs_nomes,
    "Resposta_nao_negativa": subs_nomes,
}

colunas = {
    "Known_stereotyped_groups_pt":"grupos_estereotipados",
    "grupos_estereotipados":"grupos_estereotipados",
    "Contexto_ambiguous":"Contexto_Ambiguous",
    "Questao_negativa_stereotype": "Questao_negativa"
}

for nome in nomeArqs:
    df = normalizando_dados(nome, substituicoes, colunas)
```
